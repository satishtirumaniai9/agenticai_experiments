```python
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(model="openai:gpt-5.5", tools=[get_weather],system_prompt="You are a helpful assistant that provides weather information.")

response = agent.invoke({"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]})

print(response["messages"][-1].content_blocks)
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("openai:gpt-5.5")
response =model.invoke([{"role": "user", "content": "How should we prepare for transition to a AI job after 20 years of QA experience?"}])
print(response)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage,AIMessage

messages = [
    SystemMessage(content="You are a pirate, answer every question in pirate speak."),
    HumanMessage(content="What's the capital of France?"),
    AIMessage(content="Arrr, the capital of France be Paris, matey!")
]

response = model.invoke(messages)
print(response.content)
```

```python
for chunk in model.stream([{"role": "user", "content": "How should we prepare for transition to a AI job after 20 years of QA experience?"}]):
    print(chunk.text, end="|",flush=True)
    print(type(chunk))
```

```python
from pydantic import BaseModel,Field

class Email(BaseModel):
    subject: str = Field(..., description="The subject of the email")
    body: str = Field(..., description="The body of the email")
    sender: str = Field(..., description="The sender's email address")
    recipient: str = Field(..., description="The recipient's email address")
```

```python
from pydantic import BaseModel, Field


class Email(BaseModel):
    subject: str = Field(description="The subject of the email")
    body: str = Field(description="The main content of the email")
    sender: str = Field(description="The sender's email address")
    recipient: str = Field(description="The recipient's email address")


model_with_structure = model.with_structured_output(Email)

model_with_structure_response = model_with_structure.invoke(
    [
        {
            "role": "user",
            "content": """
Generate an email with the following details:

Subject: Meeting Reminder
Body: Don't forget about our meeting tomorrow at 10 AM.
Sender: john.doe@example.com
Recipient: jane.doe@example.com
""",
        }
    ]
)

print(model_with_structure_response)
```

```python
from langchain_core.messages import (
    HumanMessage,
    AIMessage,
    ToolMessage
)

messages = [
    HumanMessage(content="What's the weather in Atlanta?")
]

# AI requests a tool
ai_message = AIMessage(
    content="",
    tool_calls=[
        {
            "name": "get_weather",
            "args": {"city": "Atlanta"},
            "id": "call_123",
            "type": "tool_call",
        }
    ],
)

messages.append(ai_message)

# Tool executes
tool_message = ToolMessage(
    content="72°F, Sunny",
    tool_call_id="call_123",
)

messages.append(tool_message)

# Send messages back to the model

model_response = model.invoke(messages)
print(model_response.content) # This will show the tool calls made by the model
```
